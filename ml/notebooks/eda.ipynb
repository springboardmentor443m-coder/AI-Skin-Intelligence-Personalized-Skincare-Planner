```python
cleaned_products = products[
    [
        "product_id",
        "product_name",
        "brand_name",
        "ingredients",
        "price_usd",
        "rating",
        "reviews",
        "loves_count",
        "highlights",
        "primary_category",
        "secondary_category",
        "tertiary_category"
    ]
]

cleaned_products.head()
```

```python
cleaned_products.isnull().sum()
```

```python
cleaned_products["primary_category"].value_counts()
```

```python
skincare_products = cleaned_products[
    cleaned_products["primary_category"] == "Skincare"
].copy()

skincare_products.shape
```

```python
skincare_products.isnull().sum()
```

```python
skincare_products.head(10)
```

```python
skincare_products["ingredients"] = skincare_products["ingredients"].fillna("Unknown")

skincare_products["highlights"] = skincare_products["highlights"].fillna("Not Available")

skincare_products["tertiary_category"] = skincare_products["tertiary_category"].fillna("Unknown")

skincare_products["rating"] = skincare_products["rating"].fillna(
    skincare_products["rating"].median()
)

skincare_products["reviews"] = skincare_products["reviews"].fillna(0)
```

```python
skincare_products.isnull().sum()
```

```python
skincare_products.to_csv("../dataset/cleaned_skincare_products.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python
import pandas as pd

df = pd.read_csv("../dataset/cleaned_skincare_products.csv")

df.head()
```

```python
df.columns
```

```python
# Create one text column containing important product information

df["combined_features"] = (
    df["product_name"].astype(str) + " " +
    df["brand_name"].astype(str) + " " +
    df["primary_category"].astype(str) + " " +
    df["secondary_category"].astype(str) + " " +
    df["tertiary_category"].astype(str) + " " +
    df["highlights"].astype(str) + " " +
    df["ingredients"].astype(str)
)

df[["product_name", "combined_features"]].head()
```

```python
df[["product_name", "combined_features"]].head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create TF-IDF vectorizer
tfidf = TfidfVectorizer(stop_words="english")

# Convert text into numerical vectors
tfidf_matrix = tfidf.fit_transform(df["combined_features"])

print("TF-IDF Matrix Shape:", tfidf_matrix.shape)
```
