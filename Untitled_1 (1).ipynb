# 1. Import Libraries

````python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
from sklearn.metrics import classification_report, confusion_matrix
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import zipfile
import shutil
````

# 2. Load Dataset

````python
# Unzip the dataset
zip_path = '/content/Skilcare dataset -20260714T070554Z-1-001 (1).zip'
extract_path = '/content/skincare_dataset'

# Create the directory if it doesn't exist
if not os.path.exists(extract_path):
    os.makedirs(extract_path)

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print(f"Dataset unzipped to: {extract_path}")

# Adjust the path to the actual dataset directory structure
# Dynamically find the main directory containing the class folders
dataset_content_dirs = [d for d in os.listdir(extract_path) if os.path.isdir(os.path.join(extract_path, d))]
if dataset_content_dirs:
    # Use the first subdirectory found directly (without stripping) to match the exact folder name
    dataset_base_path = os.path.join(extract_path, dataset_content_dirs[0])
    print(f"Detected main dataset content path: {dataset_base_path}")
else:
    dataset_base_path = extract_path # Fallback if no subdirectories are found
    print(f"No subdirectories found in {extract_path}, using {dataset_base_path} as base path.")

print(f"Using dataset path: {dataset_base_path}")

# Define image dimensions and batch size
img_height = 224 # From kernel state
img_width = 224  # From kernel state
batch_size = 32  # From kernel state

# Find all subdirectories which are likely class labels
class_labels_found = sorted([d.name for d in os.scandir(dataset_base_path) if d.is_dir()])
print(f"Found class labels: {class_labels_found}")

# If no class labels found, assume dataset_base_path is the direct parent of image files
if not class_labels_found:
    print("No subdirectories found within the detected base path. Please check dataset structure.")


# Function to create train/val/test directories and split files
def create_train_val_test_dirs(source_dir, output_base_dir, train_ratio=0.7, val_ratio=0.15, test_ratio=0.15):
    if not (train_ratio + val_ratio + test_ratio == 1.0):
        raise ValueError("Ratios must sum to 1.0")

    # Ensure a clean slate by removing existing output directory
    if os.path.exists(output_base_dir):
        shutil.rmtree(output_base_dir)
        print(f"Removed existing directory: {output_base_dir}")

    train_dir = os.path.join(output_base_dir, 'train')
    val_dir = os.path.join(output_base_dir, 'validation')
    test_dir = os.path.join(output_base_dir, 'test')

    os.makedirs(train_dir, exist_ok=True)
    os.makedirs(val_dir, exist_ok=True)
    os.makedirs(test_dir, exist_ok=True)

    for class_name in os.listdir(source_dir):
        class_path = os.path.join(source_dir, class_name)
        if os.path.isdir(class_path):
            os.makedirs(os.path.join(train_dir, class_name), exist_ok=True)
            os.makedirs(os.path.join(val_dir, class_name), exist_ok=True)
            os.makedirs(os.path.join(test_dir, class_name), exist_ok=True)

            images = [f for f in os.listdir(class_path) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
            np.random.shuffle(images)

            n_total = len(images)
            n_train = int(n_total * train_ratio)
            n_val = int(n_total * val_ratio)
            # n_test will be the rest, ensuring sum is n_total

            for i, image_name in enumerate(images):
                src_file = os.path.join(class_path, image_name)
                if i < n_train:
                    dst_file = os.path.join(train_dir, class_name, image_name)
                elif i < n_train + n_val:
                    dst_file = os.path.join(val_dir, class_name, image_name)
                else:
                    dst_file = os.path.join(test_dir, class_name, image_name)
                shutil.copy(src_file, dst_file)
    print(f"Dataset split and copied to {output_base_dir}")
    return train_dir, val_dir, test_dir

# Create the split dataset structure
split_base_dir = '/content/split_skincare_dataset'
train_data_dir, val_data_dir, test_data_dir = create_train_val_test_dirs(
    dataset_base_path, split_base_dir, train_ratio=0.7, val_ratio=0.15, test_ratio=0.15
)

print(f"Train directory: {train_data_dir}")
print(f"Validation directory: {val_data_dir}")
print(f"Test directory: {test_data_dir}")

print("\n--- Verifying Created Directory Structure ---")
def print_directory_structure(startpath, max_depth=3):
    for root, dirs, files in os.walk(startpath):
        level = root.replace(startpath, '').count(os.sep)
        if level > max_depth:
            continue
        indent = ' ' * 4 * (level)
        print(f'{indent}{os.path.basename(root)}/')
        subindent = ' ' * 4 * (level + 1)
        for f in files:
            if f.endswith(('.png', '.jpg', '.jpeg')):
                print(f'{subindent}{f}')
            elif len(files) > 3 and f == files[2]: # Limit file listing if many
                print(f'{subindent}...')
                break
        if level == max_depth and dirs:
            print(f'{subindent}...') # Indicate there are more subdirectories not shown

print_directory_structure(split_base_dir)
````

# 3. Data Augmentation

````python
# Apply data augmentation to the training set
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    zoom_range=0.15,
    width_shift_range=0.2,
    height_shift_range=0.2,
    horizontal_flip=True,
    brightness_range=[0.8, 1.2],
    fill_mode='nearest'
)

# Only rescale for validation and test sets (no augmentation)
val_datagen = ImageDataGenerator(rescale=1./255)
test_datagen = ImageDataGenerator(rescale=1./255)

# Load images from directories using the generators
# Explicitly pass class_labels_found to ensure correct class recognition
train_generator = train_datagen.flow_from_directory(
    train_data_dir,
    target_size=(img_height, img_width),
    batch_size=batch_size,
    class_mode='categorical',
    classes=class_labels_found
)

validation_generator = val_datagen.flow_from_directory(
    val_data_dir,
    target_size=(img_height, img_width),
    batch_size=batch_size,
    class_mode='categorical',
    classes=class_labels_found
)

test_generator = test_datagen.flow_from_directory(
    test_data_dir,
    target_size=(img_height, img_width),
    batch_size=batch_size,
    class_mode='categorical',
    shuffle=False, # Keep data in order for evaluation
    classes=class_labels_found
)

# Get class labels from the generator
class_labels_tl = list(train_generator.class_indices.keys())
print(f"Class labels for training: {class_labels_tl}")
print(f"Number of classes: {len(class_labels_tl)}")
````

# 4. Build MobileNetV2 Model

````python
num_classes = len(class_labels_tl)

# Load the MobileNetV2 model with pre-trained ImageNet weights
# Exclude the top classification layer
base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(img_height, img_width, 3))

# Freeze the base model layers
base_model.trainable = False

# Add custom classification layers on top of MobileNetV2
x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(128, activation='relu')(x)
predictions = Dense(num_classes, activation='softmax')(x)

model = Model(inputs=base_model.input, outputs=predictions)

# Compile the model
model.compile(optimizer=Adam(learning_rate=0.0001), loss='categorical_crossentropy', metrics=['accuracy'])

print("Model built successfully with a frozen MobileNetV2 base.")
model.summary()

# Fine-tune the last 30 layers of MobileNetV2
# Unfreeze the last 30 layers of the base model
base_model.trainable = True

# Fine-tune from this layer onwards
fine_tune_at = len(base_model.layers) - 30

# Freeze all layers before the `fine_tune_at` layer
for layer in base_model.layers[:fine_tune_at]:
    layer.trainable = False

# Recompile the model for fine-tuning with a lower learning rate
model.compile(optimizer=Adam(learning_rate=0.00001), loss='categorical_crossentropy', metrics=['accuracy'])

print(f"MobileNetV2 base model unfrozen and set to fine-tune from layer {fine_tune_at} onwards.")
model.summary()
````

# 5. Train Model

````python
epochs = 10 # You can adjust this number

# Define Callbacks
# EarlyStopping: Stop training when a monitored quantity has stopped improving.
ear_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)

# ReduceLROnPlateau: Reduce learning rate when a metric has stopped improving.
reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=5, min_lr=0.000001)

# ModelCheckpoint: Save the best model during training.
model_checkpoint = ModelCheckpoint(
    'skin_model_mobilenetv2.keras', # .keras is the recommended format
    monitor='val_accuracy', # Monitor validation accuracy
    save_best_only=True, # Save only the best model
    mode='max', # Maximize validation accuracy
    verbose=1
)

callbacks = [ear_stopping, reduce_lr, model_checkpoint]

# Train the model
history = model.fit(
    train_generator,
    epochs=epochs,
    validation_data=validation_generator,
    callbacks=callbacks
)

print("Model training complete.")
````

# 6. Evaluate Model

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Display training and validation accuracy/loss graphs
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Training and Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()

# Evaluate the model on the test set
print("Evaluating model on test set...")
test_loss, test_accuracy = model.evaluate(test_generator)
print(f"Test Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_accuracy:.4f}")

# Generate confusion matrix and classification report
print("Generating predictions for confusion matrix and classification report...")
test_generator.reset() # Ensure predictions are in order
predictions = model.predict(test_generator)
y_pred = np.argmax(predictions, axis=1)
y_true = test_generator.classes

# Get class names in the order of prediction
class_names = list(test_generator.class_indices.keys())

print("\nClassification Report:")
print(classification_report(y_true, y_pred, target_names=class_names))

# Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.show()

print("Model evaluation complete. Best model saved as skin_model_mobilenetv2.keras.")
````

# 7. Compare Images

````python
from IPython.display import display, HTML
from PIL import Image
import io
import base64

# Function to preprocess a single image
def preprocess_image(image_path, target_size=(img_height, img_width)):
    img = tf.keras.utils.load_img(image_path, target_size=target_size)
    img_array = tf.keras.utils.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0) # Create a batch
    return img_array / 255.0  # Rescale

# Function to predict skin condition and confidence
def predict_skin_condition(model, image_path, class_names):
    preprocessed_img = preprocess_image(image_path)
    predictions = model.predict(preprocessed_img)
    confidence_scores = predictions[0]
    predicted_class_index = np.argmax(confidence_scores)
    predicted_class = class_names[predicted_class_index]
    confidence = confidence_scores[predicted_class_index] * 100
    return predicted_class, confidence, confidence_scores

# Function to calculate improvement (example logic - needs domain specific rules)
def calculate_improvement(
    before_condition, before_confidence_scores,
    after_condition, after_confidence_scores,
    class_names
):
    # This is a placeholder for actual domain-specific improvement calculation.
    # For a real application, you'd define rules based on skin conditions.
    # Example: If 'Acne' confidence decreases, it's an improvement.
    # If 'Healthy' confidence increases, it's an improvement.

    improvement_percentage = 0.0
    changes = []

    # Simple example: if 'clear skin' confidence increases
    if 'clear skin' in class_names:
        clear_skin_idx = class_names.index('clear skin')
        if after_confidence_scores[clear_skin_idx] > before_confidence_scores[clear_skin_idx]:
            improvement_percentage = (after_confidence_scores[clear_skin_idx] - before_confidence_scores[clear_skin_idx]) * 100
            changes.append(f"Increased 'clear skin' confidence by {improvement_percentage:.2f}%")

    # Simple example: if a 'problem' condition confidence decreases
    problem_conditions = [c for c in class_names if c not in ['clear skin']]
    for pc in problem_conditions:
        if pc in class_names:
            pc_idx = class_names.index(pc)
            if after_confidence_scores[pc_idx] < before_confidence_scores[pc_idx]:
                diff = (before_confidence_scores[pc_idx] - after_confidence_scores[pc_idx]) * 100
                changes.append(f"Decreased '{pc}' confidence by {diff:.2f}%")
                # This is a very simplistic way to calculate, needs refinement
                improvement_percentage += diff * 0.5 # Arbitrary weighting

    return max(0, min(100, improvement_percentage)), changes

# Function to display images side-by-side with results
def display_comparison_results(
    image_before_path, image_after_path,
    pred_before, conf_before,
    pred_after, conf_after,
    improvement_percent, changes_highlight
):
    def img_to_base64(img_path):
        with open(img_path, "rb") as img_file:
            return base64.b64encode(img_file.read()).decode('utf-8')

    img_b64_before = img_to_base64(image_before_path)
    img_b64_after = img_to_base64(image_after_path)

    html_output = f"""
    <div style="display: flex; justify-content: space-around; font-family: sans-serif;">
        <div style="text-align: center;">
            <h3>Before Image</h3>
            <img src="data:image/jpeg;base64,{img_b64_before}" style="width: 250px; height: 250px; object-fit: cover; border-radius: 8px; box-shadow: 0 4px 8px rgba(0,0,0,0.1);"/>
            <p><b>Predicted Condition:</b> {pred_before}</p>
            <p><b>Confidence:</b> {conf_before:.2f}%</p>
        </div>
        <div style="text-align: center;">
            <h3>After Image</h3>
            <img src="data:image/jpeg;base64,{img_b64_after}" style="width: 250px; height: 250px; object-fit: cover; border-radius: 8px; box-shadow: 0 4px 8px rgba(0,0,0,0.1);"/>
            <p><b>Predicted Condition:</b> {pred_after}</p>
            <p><b>Confidence:</b> {conf_after:.2f}%</p>
        </div>
    </div>
    <div style="text-align: center; margin-top: 20px; padding: 20px; background-color: #f9f9f9; border-radius: 8px; box-shadow: 0 4px 8px rgba(0,0,0,0.05);">
        <h3>Comparison Results</h3>
        <p style="font-size: 1.2em; color: #28a745;"><b>Improvement:</b> {improvement_percent:.2f}%</p>
        <p><b>Highlighted Changes:</b></p>
        <ul>
            {''.join([f'<li>{change}</li>' for change in changes_highlight])}
        </ul>
    </div>
    """
    display(HTML(html_output))

# --- Example Usage ---
# First, check if specific 'before' and 'after' images were uploaded.
# You can change these filenames to match what you upload.
UPLOAD_DIR = '/content/uploaded_images'
BEFORE_IMG_NAME = 'before_skin.jpg'
AFTER_IMG_NAME = 'after_skin.jpg'

uploaded_before_path = os.path.join(UPLOAD_DIR, BEFORE_IMG_NAME)
uploaded_after_path = os.path.join(UPLOAD_DIR, AFTER_IMG_NAME)

image_before_path = None
image_after_path = None

if os.path.exists(uploaded_before_path) and os.path.exists(uploaded_after_path):
    image_before_path = uploaded_before_path
    image_after_path = uploaded_after_path
    print(f"Using uploaded images for comparison: {BEFORE_IMG_NAME} and {AFTER_IMG_NAME}")
elif len(test_generator.filenames) > 1:
    # Fallback: Use two images from the test set for demonstration
    image_before_path = os.path.join(test_data_dir, test_generator.filenames[0])
    image_after_path = os.path.join(test_data_dir, test_generator.filenames[1])
    print(f"Using demo images from test set for comparison: {os.path.basename(image_before_path)} and {os.path.basename(image_after_path)}")
else:
    print("Not enough suitable images found for comparison. Please upload or ensure test set has enough images.")


if image_before_path and image_after_path:
    # Predict conditions for selected images
    pred_before, conf_before, conf_scores_before = predict_skin_condition(model, image_before_path, class_labels_tl)
    pred_after, conf_after, conf_scores_after = predict_skin_condition(model, image_after_path, class_labels_tl)

    # Calculate improvement
    improvement_percent, changes_highlight = calculate_improvement(
        pred_before, conf_scores_before,
        pred_after, conf_scores_after,
        class_labels_tl
    )

    # Display results
    display_comparison_results(
        image_before_path, image_after_path,
        pred_before, conf_before,
        pred_after, conf_after,
        improvement_percent, changes_highlight
    )
````

## 9. Image Pipeline Workflow Diagram

To provide clarity on how images are processed, here's a conceptual workflow diagram of the AI Skin Intelligence & Personalized Skincare Planner:

```mermaid
graph TD
    A[Image Upload] --> B{Preprocessing: Resize, Normalize};
    B --> C{Skin Disease/Condition Classification (MobileNetV2)};
    C --> D{Prediction Output: Class, Confidence Score};
    D --> E{Enhanced Output Generation:
      - Skin Health Score
      - Severity Level
      - Detected Concerns
      - Recommended Ingredients
      - Recommended Products
      - Morning/Evening Routines
      - Weekly Plan};
    E --> F[Dashboard / User Interface Display];

    A -- Optional for Comparison --> G[Before/After Image Pair];
    G --> B;
    E -- Comparison Display --> F;

    style A fill:#f9f,stroke:#333,stroke-width:2px;
    style B fill:#bbf,stroke:#333,stroke-width:2px;
    style C fill:#bfb,stroke:#333,stroke-width:2px;
    style D fill:#fb9,stroke:#333,stroke-width:2px;
    style E fill:#fdd,stroke:#333,stroke-width:2px;
    style F fill:#ccf,stroke:#333,stroke-width:2px;
    style G fill:#ffb,stroke:#333,stroke-width:2px;
```

**Explanation of the Workflow:**

1.  **Image Upload:** The user uploads one or more skin images (e.g., for 'Before/After' comparison or single-image analysis).
2.  **Preprocessing:** Uploaded images are resized to the model's expected input dimensions (224x224 pixels) and normalized (pixel values scaled to 0-1).
3.  **Skin Disease/Condition Classification (MobileNetV2):** The preprocessed image is fed into the fine-tuned MobileNetV2 model.
4.  **Prediction Output:** The model outputs a predicted skin condition (e.g., 'clear skin', 'dark spots') and a confidence score for that prediction.
5.  **Enhanced Output Generation:** Based on the predicted condition and confidence, various personalized insights and recommendations are generated. This includes a derived skin health score, severity level, specific concerns, and tailored product/routine suggestions.
6.  **Dashboard / User Interface Display:** All the information (original image, predicted condition, scores, and recommendations) is presented to the user in an organized and actionable format on a dashboard or report interface.

### Why Gemini or GPT is the best choice for skincare recommendations:

For a Skin Intelligence & Personalized Skincare Planner, **Gemini** and **GPT** (specifically GPT-4V or newer) stand out as the best choices, primarily due to their **strong multimodal capabilities** and **high recommendation quality**.

1.  **Multimodal Capability:** Our application involves analyzing skin images (`Image Upload -> ... -> Skin Classification`). Gemini's native multimodal design and GPT-4V's advanced vision capabilities allow them to interpret visual cues from the 'Before' and 'After' skin images. This is crucial for *understanding* the skin condition in context and not just relying on the classification label. For instance, an LLM with strong multimodal understanding could differentiate between different types of dark spots or wrinkles based on the image, providing more precise recommendations.

2.  **Recommendation Quality & Nuance:** Both Gemini and GPT models excel at generating human-like, nuanced, and comprehensive responses. This is vital for personalized skincare recommendations that consider various factors beyond just the predicted condition. They can provide detailed explanations for why certain ingredients or routines are suggested, adapt to user queries, and offer a more engaging and informative user experience.

3.  **Medical Reasoning (with caveats):** While no LLM should replace professional medical advice, these models demonstrate better 'medical reasoning' capabilities in understanding skin concerns and suggesting appropriate (though still general) remedies. Their extensive training data includes a vast amount of health and beauty information, enabling them to formulate coherent and relevant skincare advice. The emphasis on safety in models like Claude is also a plus here, but its multimodal capabilities are currently less developed than Gemini or GPT-4V.

4.  **API Support & Integration:** Both OpenAI's and Google's ecosystems offer robust API support, making integration into a larger application (like our Colab notebook) relatively straightforward. This ensures scalability and ease of development.

In summary, while Llama 3 offers flexibility and cost-effectiveness for custom deployments, its current lack of strong multimodal support makes it less ideal for an image-centric application like ours. Claude is strong in safety and reasoning, but Gemini and GPT-4V's superior multimodal understanding combined with their high-quality recommendation generation make them the top contenders for providing an intelligent and personalized skincare planning experience.
